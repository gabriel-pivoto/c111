## Questão 1

Crie um programa que leia seu nome completo e mostre:

- Seu nome com todas as letras maiúsculas;
- Seu nome com todas as letras minúsculas;
- Quantas letras ao todo tem seu nome;
- E como seria se trocássemos seu último nome para "do Inatel".

```python
full_name = input('Digite seu nome completo: ').strip()

print('Maiúsculas:', full_name.upper())
print('Minúsculas:', full_name.lower())

letters = len(full_name.replace(' ', ''))
print('Quantidade de letras (sem espaços):', letters)
print('Quantidade de caracteres (com espaços):', len(full_name))

parts = full_name.split()
if len(parts) > 1:
    new_name = ' '.join(parts[:-1]) + ' do Inatel'
else:
    new_name = full_name + ' do Inatel'

print('Com o último nome trocado:', new_name)
```

## Questão 2

Mostre a tabuada de um número que o usuário escolher dentro de um intervalo específico também escolhido por ele.

```python
number = int(input('Digite o número da tabuada: '))
start = int(input('Digite o início do intervalo: '))
end = int(input('Digite o fim do intervalo: '))

if start > end:
    start, end = end, start

print(f'\nTabuada do {number} de {start} até {end}:')
for multiplier in range(start, end + 1):
    print(f'{number} x {multiplier} = {number * multiplier}')
```

## Questão 3

Faça um programa que leia o sexo de uma pessoa e diga se ela é homem (caso seja digitado M) ou mulher (caso seja digitado F). Caso seja digitado algo inválido, continue perguntando até que o usuário entre com um sexo válido.

```python
gender = input('Digite seu sexo [M/F]: ').strip().upper()

while gender not in ('M', 'F'):
    print('Valor inválido! Digite apenas M ou F.')
    gender = input('Digite seu sexo [M/F]: ').strip().upper()

if gender == 'M':
    print('Você é homem.')
else:
    print('Você é mulher.')
```

## Questão 4

Desenvolva um script que pergunte a distância de uma viagem em Km. Calcule o preço da passagem, cobrando R$0.50 por Km para viagens até 200Km e R$0.45 para viagens mais longas.

```python
distance = float(input('Qual a distância da viagem (em Km)? '))

if distance <= 200:
    price_per_km = 0.50
else:
    price_per_km = 0.45

price = distance * price_per_km

print(f'Distância: {distance:.1f} Km')
print(f'Preço por Km: R${price_per_km:.2f}')
print(f'Preço da passagem: R${price:.2f}')
```

## Questão 5

Faça um programa que leia um número entre 1000 e 9999 e mostre na tela:

- Qual o número da unidade;
- Número da dezena;
- Número da centena;
- E número do milhar.

```python
number = int(input('Digite um número entre 1000 e 9999: '))

while number < 1000 or number > 9999:
    print('Número fora do intervalo! Tente novamente.')
    number = int(input('Digite um número entre 1000 e 9999: '))

units = number % 10
tens = number // 10 % 10
hundreds = number // 100 % 10
thousands = number // 1000 % 10

print(f'\nAnalisando o número {number}:')
print('Unidade:', units)
print('Dezena:', tens)
print('Centena:', hundreds)
print('Milhar:', thousands)
```

## Questão 6

Peça ao usuário para entrar com um número decimal. Em seguida, aplique e mostre o resultado:

- Da raiz quadrada deste número;
- Função teto;
- Função chão;
- Sua parte inteira.

```python
import math

number = float(input('Digite um número decimal: '))

print(f'Número digitado: {number}')

if number >= 0:
    print(f'Raiz quadrada: {math.sqrt(number):.4f}')
else:
    print('Raiz quadrada: não existe raiz real para número negativo.')

print('Função teto (ceil):', math.ceil(number))
print('Função chão (floor):', math.floor(number))
print('Parte inteira (trunc):', math.trunc(number))
```

## Questão 7

Faça um programa que leia uma palavra. Esse programa deve percorrer a palavra e imprimir cada letra em maiúsculo, uma letra por linha. Ao final, deverá informar quantas vogais a palavra tem e se a letra "a" está presente nela.

```python
word = input('Digite uma palavra: ').strip()

vowel_count = 0
for letter in word:
    print(letter.upper())
    if letter.lower() in 'aeiou':
        vowel_count += 1

print(f'\nA palavra "{word}" tem {vowel_count} vogal(is).')

if 'a' in word.lower():
    print('A letra "a" está presente na palavra.')
else:
    print('A letra "a" NÃO está presente na palavra.')
```
