## Questão 1

Carregue o Dataset `paises.csv` e em seguida mostre:

- Quais são os países da OCEANIA;
- Quantos países são da OCEANIA.

O arquivo usa vírgula como separador decimal, então `decimal=','` faz o pandas ler essas colunas como número em vez de texto. Os nomes em `Country` e `Region` vêm com espaços sobrando no final, e `str.strip()` limpa isso antes de qualquer filtro.

```python
import pandas as pd

df = pd.read_csv('paises.csv', decimal=',')

df['Country'] = df['Country'].str.strip()
df['Region'] = df['Region'].str.strip()

oceania = df[df['Region'].str.contains('OCEANIA')]

print('Países da OCEANIA:')
print(oceania['Country'].to_string(index=False))

print(f'\nQuantos países são da OCEANIA: {len(oceania)}')
```

## Questão 2

Encontre o nome e a região do país que possui a maior população segundo este Dataset.

`idxmax()` devolve o índice da linha de maior valor, e `loc` usa esse índice para trazer a linha inteira.

```python
mais_populoso = df.loc[df['Population'].idxmax()]

print(f"País com a maior população: {mais_populoso['Country']}")
print(f"Região: {mais_populoso['Region']}")
print(f"População: {mais_populoso['Population']:,}")
```

## Questão 3

Agrupe os países por Regiões. Em seguida, mostre a média de alfabetização (Literacy (%)) de cada região do planeta.

18 países estão sem o dado de alfabetização. O `mean()` ignora esses valores em vez de devolver NaN para a região toda, então cada média sai calculada sobre os países que têm o dado.

```python
media_alfabetizacao = df.groupby('Region')['Literacy (%)'].mean()

print('Média de alfabetização (%) de cada região:')
print(media_alfabetizacao.sort_values(ascending=False).round(2))
```

## Questão 4

Busque o nome de todos os países do Dataset que não possuem costa marítima (Coastline (coast/area ratio) == 0) e guarde-os em um novo arquivo chamado `noCoast.csv`.

`index=False` evita gravar a coluna de índice do pandas como se fosse mais um dado do arquivo.

```python
sem_costa = df[df['Coastline (coast/area ratio)'] == 0]

print(f'Países sem saída para o mar: {len(sem_costa)}')
print(sem_costa['Country'].to_string(index=False))

sem_costa.to_csv('noCoast.csv', index=False)
print(f'\nArquivo noCoast.csv gravado com {len(sem_costa)} linhas')
```

## Questão 5

Faça uma função que receba a taxa de mortalidade de cada país (Deathrate) e retorne o texto 'Balanced' caso o valor seja < 9 e 'Urgent' caso contrário. Em seguida, crie um campo novo no Dataset chamado 'Humanitarian Help' que receba estes valores para cada país. No final, mostre o Dataset para verificar se a inserção da nova coluna foi feita com sucesso.

Quatro países estão sem o Deathrate preenchido. Como `NaN < 9` é falso, eles cairiam em 'Urgent' sem nunca terem sido avaliados, então a função trata esse caso à parte.

```python
def ajuda_humanitaria(taxa):
    if pd.isna(taxa):
        return 'Sem dados'
    return 'Balanced' if taxa < 9 else 'Urgent'


df['Humanitarian Help'] = df['Deathrate'].apply(ajuda_humanitaria)

print('Primeiras 10 linhas do Dataset com a nova coluna:')
print(df[['Country', 'Region', 'Deathrate', 'Humanitarian Help']].head(10))

print('\nContagem de cada valor da nova coluna:')
print(df['Humanitarian Help'].value_counts())

print(f'\nO Dataset agora tem {len(df.columns)} colunas e a última é "{df.columns[-1]}"')
```
