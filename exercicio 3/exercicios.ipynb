## Questão 1

Crie uma lista preenchida com os 5 primeiros colocados de um Campeonato de Futebol na ordem de colocação. Depois mostre:

- Apenas os 3 primeiros colocados;
- Os últimos 2 colocados;
- Uma lista com os times em ordem alfabética;
- Em que posição da tabela se encontra o Barcelona.

```python
teams = ['Real Madrid', 'Barcelona', 'Atlético de Madrid', 'Sevilla', 'Villarreal']

print('Tabela completa:', teams)
print('Os 3 primeiros colocados:', teams[:3])
print('Os 2 últimos colocados:', teams[-2:])
print('Times em ordem alfabética:', sorted(teams))

if 'Barcelona' in teams:
    position = teams.index('Barcelona') + 1
    print(f'O Barcelona está na {position}ª posição da tabela.')
else:
    print('O Barcelona não está na tabela.')
```

## Questão 2

Crie três conjuntos, um para cada loja. Identifique quais modelos de smartphones cada uma delas vende. Em seguida, mostre quantos modelos no total você terá opção de comprar se visitá-las e quais modelos se encontram disponíveis em ambas as lojas.

```python
store_a = {'iPhone 15', 'Galaxy S24', 'Xiaomi 14', 'Moto G84'}
store_b = {'Galaxy S24', 'Xiaomi 14', 'Pixel 8', 'Moto G84'}
store_c = {'iPhone 15', 'Pixel 8', 'Galaxy A55', 'Galaxy S24'}

print('A loja A vende:', store_a)
print('A loja B vende:', store_b)
print('A loja C vende:', store_c)

all_models = store_a | store_b | store_c
print(f'\nVisitando as 3 lojas você terá {len(all_models)} modelos como opção de compra:')
print(all_models)

print('\nModelos disponíveis nas lojas A e B:', store_a & store_b)
print('Modelos disponíveis nas lojas A e C:', store_a & store_c)
print('Modelos disponíveis nas lojas B e C:', store_b & store_c)
print('Modelos disponíveis nas três lojas:', store_a & store_b & store_c)
```

## Questão 3

Faça um programa que leia o nome e a média de um aluno e guarde em um dicionário. Em seguida, a partir da média (para ser aprovado deve ter média >=50), gere a situação final do aluno ('AP' ou 'RP'), que também deve ser guardada neste dicionário. No final, mostre todo o conteúdo deste dicionário.

```python
student = {}

student['nome'] = input('Digite o nome do aluno: ').strip()
student['media'] = float(input('Digite a média do aluno: '))

if student['media'] >= 50:
    student['situacao'] = 'AP'
else:
    student['situacao'] = 'RP'

print('\nDados do aluno:')
for key, value in student.items():
    print(f'{key}: {value}')
```

## Questão 4

Faça um programa que leia o nome e o peso de 3 pessoas e no final mostre o nome da pessoa mais pesada e a mais leve.

```python
heaviest_name = ''
heaviest_weight = 0.0
lightest_name = ''
lightest_weight = 0.0

for position in range(1, 4):
    name = input(f'Digite o nome da {position}ª pessoa: ').strip()
    weight = float(input(f'Digite o peso de {name} (em Kg): '))

    if position == 1:
        heaviest_name, heaviest_weight = name, weight
        lightest_name, lightest_weight = name, weight
    else:
        if weight > heaviest_weight:
            heaviest_name, heaviest_weight = name, weight
        if weight < lightest_weight:
            lightest_name, lightest_weight = name, weight

print(f'\nA pessoa mais pesada é {heaviest_name}, com {heaviest_weight:.1f} Kg.')
print(f'A pessoa mais leve é {lightest_name}, com {lightest_weight:.1f} Kg.')
```

## Questão 5

Desenvolva um programa que leia o nome, idade e sexo de n pessoas. No final, mostre:

- A média de idade do grupo;
- Quantas mulheres têm menos de 20 anos.

**Dica:** Em Python, os operadores booleanos básicos são **and**, **or** e **not**.

```python
total_people = int(input('Quantas pessoas serão cadastradas? '))

age_sum = 0
young_women = 0

for position in range(1, total_people + 1):
    name = input(f'\nDigite o nome da {position}ª pessoa: ').strip()
    age = int(input(f'Digite a idade de {name}: '))
    gender = input(f'Digite o sexo de {name} [M/F]: ').strip().upper()

    while gender not in ('M', 'F'):
        print('Valor inválido! Digite apenas M ou F.')
        gender = input(f'Digite o sexo de {name} [M/F]: ').strip().upper()

    age_sum += age

    if gender == 'F' and age < 20:
        young_women += 1

average_age = age_sum / total_people

print(f'\nA média de idade do grupo é {average_age:.1f} anos.')
print(f'{young_women} mulher(es) tem menos de 20 anos.')
```

## Questão 7

Usando uma lista de ingredientes já preenchida para uma receita de bolo. Crie dois conjuntos representando os ingredientes que duas pessoas diferentes já têm em casa. Mostre quais ingredientes da receita ainda faltam comprar.

**Dica:** Use diferença de conjuntos entre a lista de ingredientes e a união dos dois conjuntos.

```python
recipe = ['farinha', 'fermento', 'ovos', 'leite', 'manteiga']

person_a = {'farinha', 'ovos', 'manteiga'}
person_b = {'açúcar', 'ovos'}

print('Ingredientes da receita:', recipe)
print('A primeira pessoa já tem:', person_a)
print('A segunda pessoa já tem:', person_b)

available = person_a | person_b
print('\nJuntando as duas pessoas, já temos:', available)

missing = set(recipe) - available
print('Ainda faltam comprar:', missing)
```

## Questão 8

Crie um dicionário para armazenar os dados de um produto (nome, preço e quantidade em estoque). Peça ao usuário os dados de 3 produtos diferentes e guarde cada dicionário em uma lista. No final, percorra a lista e mostre, para cada produto, seu nome e o valor total em estoque (preço × quantidade).

```python
products = []

for position in range(1, 4):
    name = input(f'\nDigite o nome do {position}º produto: ').strip()
    price = float(input(f'Digite o preço de {name}: R$'))
    quantity = int(input(f'Digite a quantidade de {name} em estoque: '))

    products.append({'nome': name, 'preco': price, 'quantidade': quantity})

print('\nValor total em estoque de cada produto:')
for product in products:
    total = product['preco'] * product['quantidade']
    print(f'{product["nome"]}: R${total:.2f}')
```
