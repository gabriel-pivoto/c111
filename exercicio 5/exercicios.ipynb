# Análise do dataset `space.csv` com NumPy

O arquivo é separado por `;` e possui as colunas:

| Índice | Coluna |
| --- | --- |
| 0 | Num |
| 1 | Company Name |
| 2 | Location |
| 3 | Datum |
| 4 | Detail |
| 5 | Status Rocket |
| 6 | Cost |
| 7 | Status Mission |

A leitura é feita com `np.genfromtxt`, mantendo tudo como texto para depois converter
apenas a coluna de custo para número.

```python
import numpy as np

data = np.genfromtxt(
    'space.csv',
    delimiter=';',
    dtype=str,
    encoding='utf-8-sig',
    skip_header=1,
)

companies = data[:, 1]
locations = data[:, 2]
rocket_status = data[:, 5]
mission_status = data[:, 7]

costs = np.char.strip(data[:, 6])
costs = np.char.replace(costs, ',', '').astype(float)

print('Total de missões no dataset:', data.shape[0])
print('Colunas lidas:', data.shape[1])
print('\nPrimeira missão:', data[0])
```

## Questão 1

Apresente a porcentagem de missões que deram certo.

```python
total_missions = mission_status.size
successful = np.count_nonzero(mission_status == 'Success')

success_rate = successful / total_missions * 100

print(f'Missões bem sucedidas: {successful} de {total_missions}')
print(f'Porcentagem de missões que deram certo: {success_rate:.2f}%')
```

## Questão 2

Qual a média de gastos de uma missão espacial se baseando em missões que possuam valores
disponíveis (> 0)?

```python
available_costs = costs[costs > 0]

average_cost = available_costs.mean()

print(f'Missões com custo disponível: {available_costs.size}')
print(f'Média de gastos de uma missão: {average_cost:.2f} milhões de dólares')
print(f'Missão mais barata: {available_costs.min():.2f} milhões de dólares')
print(f'Missão mais cara: {available_costs.max():.2f} milhões de dólares')
```

## Questão 3

Encontre quantas missões espaciais neste Dataset foram realizadas pelos Estados Unidos (EUA).

```python
usa_missions = np.char.endswith(np.char.strip(locations), 'USA')

usa_total = np.count_nonzero(usa_missions)

print(f'Missões realizadas nos Estados Unidos: {usa_total}')
print(f'Isso representa {usa_total / total_missions * 100:.2f}% de todas as missões.')
```

## Questão 4

Encontre qual foi a missão mais cara realizada pela empresa "SpaceX".

```python
spacex = companies == 'SpaceX'

spacex_costs = np.where(spacex, costs, -1)
most_expensive = spacex_costs.argmax()

print('Missão mais cara da SpaceX:')
print('Detalhe:', data[most_expensive, 4])
print('Local:', data[most_expensive, 2])
print('Data:', data[most_expensive, 3])
print('Status da missão:', data[most_expensive, 7])
print(f'Custo: {costs[most_expensive]:.2f} milhões de dólares')
```

## Questão 5

Mostre o nome das empresas que já realizaram missões espaciais, junto com suas respectivas
quantidades de missões (use o for no final para mostrar as informações).

```python
company_names, mission_counts = np.unique(companies, return_counts=True)

order = np.argsort(mission_counts)[::-1]
company_names = company_names[order]
mission_counts = mission_counts[order]

print(f'{company_names.size} empresas já realizaram missões espaciais:\n\n')

for name, count in zip(company_names, mission_counts):
    print(f'{name}: {count} missão(ões)')
```

## Questão 6

Qual a porcentagem de missões realizadas com foguetes cujo status é "StatusRetired"
(coluna Status Rocket)?

```python
retired = np.count_nonzero(rocket_status == 'StatusRetired')
active = np.count_nonzero(rocket_status == 'StatusActive')

retired_rate = retired / total_missions * 100

print(f'Missões com foguetes aposentados (StatusRetired): {retired}')
print(f'Missões com foguetes ativos (StatusActive): {active}')
print(f'Porcentagem de missões com foguetes StatusRetired: {retired_rate:.2f}%')
```
