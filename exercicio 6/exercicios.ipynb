## Questão 1

Crie duas Series com os seguintes valores:

- `seriesAno1`: {'Java': 16.25, 'C': 16.04, 'Python': 9.85}
- `seriesAno2`: {'C': 16.21, 'Python': 12.12, 'Java': 11.68}

````python
import pandas as pd

seriesAno1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
seriesAno2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})

print('seriesAno1 (ano 1):')
print(seriesAno1)
print('\nseriesAno2 (ano 2):')
print(seriesAno2)
````

## Questão 2

Os valores das Series criadas na Questão 1 representam as fatias de mercado (porcentagem) de 3
linguagens de programação populares em dois anos consecutivos. Para cada ano, apresente a
porcentagem total que elas juntas representam no mercado.

````python
total_ano1 = seriesAno1.sum()
total_ano2 = seriesAno2.sum()

print(f'Ano 1: as três linguagens juntas representam {total_ano1:.2f}% do mercado')
print(f'Ano 2: as três linguagens juntas representam {total_ano2:.2f}% do mercado')
print(f'\nO restante do mercado ficou com {100 - total_ano1:.2f}% no ano 1 e {100 - total_ano2:.2f}% no ano 2')
````

## Questão 3

Apresente o crescimento/declínio no mercado de cada linguagem do primeiro ano para o segundo ano.

As duas Series estão em ordens diferentes, mas o pandas alinha os valores pelo índice antes de fazer
a conta, então a subtração já compara cada linguagem com ela mesma.

````python
diferenca = seriesAno2 - seriesAno1
variacao = diferenca / seriesAno1 * 100

comparacao = pd.DataFrame({
    'Ano 1': seriesAno1,
    'Ano 2': seriesAno2,
    'Diferença (p.p.)': diferenca.round(2),
    'Variação (%)': variacao.round(2),
})

print(comparacao)

for linguagem, taxa in variacao.items():
    situacao = 'crescimento' if taxa > 0 else 'declínio'
    print(f'{linguagem}: {situacao} de {abs(taxa):.2f}%')
````

## Questão 4

Baseado nos resultados da Questão 3, mostre apenas os dados das linguagens que tiveram crescimento.

````python
crescimento = variacao[variacao > 0]

print('Linguagens que tiveram crescimento:')
print(crescimento.round(2))
````

## Questão 5

Se estas porcentagens de crescimento/declínio se mantivessem iguais para os próximos 2 anos, qual
seria a linguagem mais popular?

A mesma taxa da Questão 3 é aplicada duas vezes sobre a fatia do ano 2, chegando no ano 4. No final,
`nlargest(1)` devolve de uma vez a label e o maior valor da Series.

````python
taxa = variacao / 100

ano3 = seriesAno2 * (1 + taxa)
ano4 = ano3 * (1 + taxa)

projecao = pd.DataFrame({
    'Ano 2': seriesAno2,
    'Ano 3': ano3.round(2),
    'Ano 4': ano4.round(2),
})
print(projecao)

mais_popular = ano4.nlargest(1)
print('\nResultado do nlargest(1):')
print(mais_popular)
print(f'\nApós 2 anos a linguagem mais popular seria {mais_popular.index[0]} com {mais_popular.iloc[0]:.2f}% do mercado')
````

calcule a média dos elementos da coluna X que são menores que 30.

````python
import numpy as np

df = pd.DataFrame(
    index=['A', 'B', 'C', 'D', 'E'],
    columns=['W', 'X', 'Y', 'Z'],
    data=np.random.randint(1, 50, [5, 4])
)

print('DataFrame:')
print(df)

menores = df['X'][df['X'] < 30]

print('\nElementos da coluna X menores que 30:')
print(menores)
print(f'\nMédia desses elementos: {menores.mean():.2f}')
````

## Questão 7

Utilizando do mesmo DataFrame, apresente a média dos elementos da linha D usando a função `loc()`
como base e a soma dos elementos da linha E usando a função `iloc()` como base.

````python
linha_d = df.loc['D']
linha_e = df.iloc[4]

print('Linha D (loc):')
print(linha_d)
print(f'\nMédia da linha D: {linha_d.mean():.2f}')

print('\nLinha E (iloc):')
print(linha_e)
print(f'\nSoma da linha E: {linha_e.sum()}')
````

## Questão 8

Faça um Slicing na matriz mostrando apenas as linhas A, C e E junto com as colunas X e Y. Em
seguida, mostre qual seria a soma dos elementos de cada uma destas linhas e cada uma destas colunas.

````python
recorte = df.loc[['A', 'C', 'E'], ['X', 'Y']]

print('Slicing das linhas A, C, E com as colunas X e Y:')
print(recorte)

print('\nSoma dos elementos de cada linha:')
print(recorte.sum(axis=1))

print('\nSoma dos elementos de cada coluna:')
print(recorte.sum(axis=0))
````
